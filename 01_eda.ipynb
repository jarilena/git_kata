```python
from utils import load_data, clean_data

# Load the raw Titanic dataset (keep original rows for summary plots)
titanic = load_data()

# Prepare a lighter `titanic_clean` that keeps only rows with an age value
# (so histograms and age-specific analyses work even if full `clean_data`
# would drop rows for other missing fields)
titanic_clean = load_data()
titanic_clean = titanic_clean.dropna(subset=['age'])

# Data loaded successfully from data/titanic.csv
titanic.head(10)
```

```python
import matplotlib.pyplot as plt

pclass_counts = titanic["pclass"].value_counts().sort_index()
fig, ax = plt.subplots(figsize=(8, 5))
pclass_counts.plot(kind="bar", ax=ax)
ax.set_title("Passenger Class Distribution")
ax.set_xlabel("Passenger Class")
ax.set_ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Drop missing ages to avoid errors
ages = titanic_clean["age"].dropna()

plt.figure(figsize=(8, 5))
plt.hist(ages, bins=20, edgecolor="black")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age distribution of passengers")
plt.grid(axis="y", alpha=0.3)
plt.show()
```
